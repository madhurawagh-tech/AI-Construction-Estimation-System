```python
import pandas as pd
df = pd.read_csv("road_construction_dataset.csv")
df.head()
```

```python
df.info()
df.columns
print("Checking null values",df.isnull().sum()) #Checking null values
print("Duplicates:",df.duplicated().sum()) #Checking duplicates
```

```python
X = df[['Road_Type','Road_Length_km','Road_Width_m','Road_Thickness_m','Number_of_Lanes','Soil_Type','Construction_Quality']]
Y = df[['Aggregate_Required_m3','Sand_Required_m3','Cement_Required_tonnes','Bitumen_Required_tonnes','Steel_Required_tonnes','Material_Cost','Labour_Cost','Total_Construction_Cost']]
```

```python
X_encoded = pd.get_dummies(X,columns=['Road_Type', 'Soil_Type', 'Construction_Quality'])

X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X_encoded, Y, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=100, random_state=42)

model.fit(X_train, Y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred)
```

```python
import numpy as np
from sklearn.metrics import  r2_score
r2 = r2_score(Y_test, y_pred)
print("R2 Score:", r2)
```
